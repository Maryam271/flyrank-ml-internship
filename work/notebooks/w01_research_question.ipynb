## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

**Quick look at the data**

The starter dataset contains **30,000 pages and 44 columns**. The correlation between `search_volume` and `impressions_90d` was **0.001**, which is very close to zero. CTR also varied substantially by position tier: the mean CTR was **0.3548 for page 1** compared with **0.0554 for deep positions**. These observed differences make content and search opportunity worth investigating further, while showing that search volume alone is not enough to explain page exposure.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("content_refresh_anonymized.csv")

print(f"Rows: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]}")

corr = df["search_volume"].corr(df["impressions_90d"])
print(f"Search volume vs impressions_90d correlation: {corr:.3f}")

visible = df[df["impressions_90d"] >= 100]
ctr_by_pos = visible.groupby("position_tier")["ctr"].mean().sort_values(ascending=False)

print("\nMean CTR by position tier:")
print(ctr_by_pos.round(4).to_string())
```
